## Hashing

Hashing is a technique or process of mapping data of arbitrary size to fixed-size values. The values generated by a hash function are called hash values, hash codes, digests, or simply hashes.

### Key Characteristics:
*   **Efficiency**: Hash functions aim to map data quickly.
*   **Determinism**: The same input must always produce the same hash value.
*   **Uniform Distribution**: A good hash function distributes hash values uniformly across the output range to minimize collisions.

**Purpose**: Hashing is primarily used for efficient data retrieval, data integrity checks, and cryptography.

### Hash Table / Dictionary in Python

A Hash Table is a data structure that implements an associative array abstract data type, a structure that can map keys to values. It uses a hash function to compute an index into an array of buckets or slots, from which the desired value can be found.

In Python, the built-in `dict` (dictionary) type is implemented as a hash table. It provides average O(1) time complexity for insertion, deletion, and lookup operations, making it extremely efficient for these tasks.

```python
print("Python dictionary (hash table) example:")
my_dict = {"apple": 1, "banana": 2, "cherry": 3}

# Insertion (average O(1))
my_dict["date"] = 4
print(f"After insertion: {my_dict}")

# Lookup (average O(1))
print(f"Value for 'banana': {my_dict['banana']}")

# Deletion (average O(1))
del my_dict["apple"]
print(f"After deletion: {my_dict}")

# Checking for key existence (average O(1))
print(f"Is 'cherry' in dict? {'cherry' in my_dict}")
```

### Collision Handling

A **collision** occurs when two different keys hash to the same index in the hash table. Since each slot can only hold one item, a strategy is needed to resolve these collisions.

Python's dictionary handles collisions internally. Common strategies used in hash table implementations include:

1.  **Separate Chaining**: Each slot in the hash table points to a linked list (or another data structure) that stores all the key-value pairs that hash to that index.
2.  **Open Addressing**: If a collision occurs, the system probes for the next available empty slot in the hash table. Common probing methods include linear probing, quadratic probing, and double hashing.

### Applications of Hashing

Hashing is fundamental to many algorithms and data structures. Let's look at a few common applications.

#### 1. Counting Frequencies

Hashing (via dictionaries in Python) is ideal for counting the occurrences of items in a collection, as it provides fast lookups and updates.

```python
print("Application: Counting Frequencies")

items = ["apple", "banana", "apple", "orange", "banana", "apple"]

# Using a dictionary to count frequencies
frequency_map = {}
for item in items:
    frequency_map[item] = frequency_map.get(item, 0) + 1

print(f"Original items: {items}")
print(f"Item frequencies: {frequency_map}")

# Using Python's collections.Counter (which also uses a hash table internally)
from collections import Counter
counted_items = Counter(items)
print(f"Item frequencies (using Counter): {counted_items}")
```

#### 2. Two Sum Problem

The Two Sum problem is a classic interview question where you need to find two numbers in an array that add up to a specific target. Hashing provides a highly efficient solution.

```python
print("Application: Two Sum Problem")

def two_sum(nums, target):
    # Use a dictionary to store numbers and their indices
    num_map = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_map: # Check if complement exists in map (O(1) average)
            return [num_map[complement], i]
        num_map[num] = i # Add current number and its index to map
    return [] # No solution found

nums1 = [2, 7, 11, 15]
target1 = 9
print(f"Numbers: {nums1}, Target: {target1}, Result: {two_sum(nums1, target1)}") # Expected: [0, 1]

nums2 = [3, 2, 4]
target2 = 6
print(f"Numbers: {nums2}, Target: {target2}, Result: {two_sum(nums2, target2)}") # Expected: [1, 2]

nums3 = [3, 3]
target3 = 6
print(f"Numbers: {nums3}, Target: {target3}, Result: {two_sum(nums3, target3)}") # Expected: [0, 1]
```
