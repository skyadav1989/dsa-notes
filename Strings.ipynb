## Python Strings: An Overview

Strings are sequences of characters, used for storing text information. They are immutable, meaning once created, their content cannot be changed.

### 1. Creating Strings

Strings can be created using single quotes `''`, double quotes `""`, or triple quotes `''' '''` or `""" """` for multi-line strings.

```python
# Using single quotes
my_string_single = 'Hello, Python!'
print(f"Single quotes: {my_string_single}")

# Using double quotes
my_string_double = "Welcome to Colab!"
print(f"Double quotes: {my_string_double}")

# Using triple quotes for multi-line strings
my_string_multi = '''This is a multi-line string.
It can span several lines.'''
print(f"Triple quotes (multi-line):\n{my_string_multi}")
```

### 2. Basic String Operations

Strings support several basic operations like concatenation (joining), repetition, and finding their length.

#### Concatenation (`+`) and Repetition (`*`)

The `+` operator joins two or more strings, while the `*` operator repeats a string a specified number of times.

```python
str1 = 'Python'
str2 = 'Programming'

# Concatenation
full_string = str1 + ' ' + str2
print(f"Concatenation: {full_string}")

# Repetition
repeated_string = 'abc' * 3
print(f"Repetition: {repeated_string}")
```

#### String Length (`len()`)

The `len()` function returns the number of characters in a string.

```python
my_string = "Data Science"
length = len(my_string)
print(f"The length of '{my_string}' is: {length}")
```

### 3. Accessing Characters (Indexing and Slicing)

Individual characters or substrings can be accessed using indexing and slicing.

#### Indexing

Characters are accessed by their position (index). Python uses zero-based indexing. Negative indices count from the end of the string.

```python
my_string = "Jupyter Notebook"

# Positive indexing
first_char = my_string[0]
fifth_char = my_string[4]
print(f"First character: {first_char}")
print(f"Fifth character: {fifth_char}")

# Negative indexing
last_char = my_string[-1]
second_to_last_char = my_string[-2]
print(f"Last character: {last_char}")
print(f"Second to last character: {second_to_last_char}")
```

#### Slicing

Slicing extracts a portion (substring) of a string. The syntax is `[start:end:step]`.
- `start`: The starting index (inclusive). If omitted, defaults to 0.
- `end`: The ending index (exclusive). If omitted, defaults to the end of the string.
- `step`: The step size. If omitted, defaults to 1.

```python
my_string = "Artificial Intelligence"

# Slice from index 0 to 9 (exclusive)
slice1 = my_string[0:10]
print(f"Slice [0:10]: {slice1}")

# Slice from index 11 to the end
slice2 = my_string[11:]
print(f"Slice [11:]: {slice2}")

# Slice from beginning to index 10 (exclusive)
slice3 = my_string[:10]
print(f"Slice [:10]: {slice3}")

# Full string copy
slice4 = my_string[:]
print(f"Slice [:]: {slice4}")

# Slice with a step of 2
slice5 = my_string[0:20:2]
print(f"Slice [0:20:2]: {slice5}")

# Reverse string using slicing
reversed_string = my_string[::-1]
print(f"Reversed string: {reversed_string}")
```

### 4. Common String Methods (Functions)

Python strings come with a rich set of built-in methods for various manipulations.

#### Case Conversion Methods

These methods change the case of characters in a string.

```python
text = "pYtHoN PrOgRaMmInG"

print(f"Original: {text}")
print(f".upper(): {text.upper()}")         # Converts to uppercase
print(f".lower(): {text.lower()}")         # Converts to lowercase
print(f".capitalize(): {text.capitalize()}") # Capitalizes first character, rest lowercase
print(f".title(): {text.title()}")           # Capitalizes first character of each word
print(f".swapcase(): {text.swapcase()}")     # Swaps case of all characters
```

#### Searching and Replacing Methods

These methods help find substrings or replace parts of a string.

```python
sentence = "I love Python programming. Python is fun."

# .find(substring) - returns the lowest index where the substring is found (-1 if not found)
print(f"'Python' found at index: {sentence.find('Python')}")
print(f"'java' found at index: {sentence.find('java')}")

# .index(substring) - similar to find, but raises ValueError if not found
try:
    print(f"'programming' found at index: {sentence.index('programming')}")
except ValueError as e:
    print(f"Error: {e}")

# .replace(old, new, count) - replaces occurrences of old with new
new_sentence = sentence.replace('Python', 'Java')
print(f"Replaced all 'Python' with 'Java': {new_sentence}")

new_sentence_limited = sentence.replace('Python', 'Java', 1)
print(f"Replaced first 'Python' with 'Java': {new_sentence_limited}")

# .count(substring) - returns the number of non-overlapping occurrences
print(f"'Python' count: {sentence.count('Python')}")
```

#### Splitting and Joining Methods

- `.split(separator)`: Divides a string into a list of substrings based on a delimiter.
- `.join(iterable)`: Joins elements of an iterable (e.g., list of strings) into a single string using the string itself as a separator.

```python
data = "apple,banana,orange,grape"

# Split by comma
fruits_list = data.split(',')
print(f"Split by comma: {fruits_list}")

words = "Hello world Python"
# Split by whitespace (default)
word_list = words.split()
print(f"Split by whitespace: {word_list}")

# Join a list of strings
new_data = '-'.join(fruits_list)
print(f"Joined with hyphen: {new_data}")

sentence_parts = ['This', 'is', 'a', 'sentence.']
joined_sentence = ' '.join(sentence_parts)
print(f"Joined with space: {joined_sentence}")
```

#### Stripping Whitespace Methods

These methods remove leading/trailing whitespace characters (spaces, tabs, newlines).

```python
padded_text = "   Hello, World!   \n"

print(f"Original (with quotes to show padding): '{padded_text}'")
print(f".strip(): '{padded_text.strip()}'")     # Removes leading/trailing whitespace
print(f".lstrip(): '{padded_text.lstrip()}'")   # Removes leading whitespace
print(f".rstrip(): '{padded_text.rstrip()}'")   # Removes trailing whitespace
```

#### Checking String Content (Boolean Methods)

These methods return `True` or `False` based on the content of the string.

```python
s1 = "12345"
s2 = "HelloWorld"
s3 = "Hello World"
s4 = "   \t\n"

print(f"'{s1}'.isdigit(): {s1.isdigit()}")         # All characters are digits
print(f"'{s2}'.isalpha(): {s2.isalpha()}")         # All characters are alphabetic
print(f"'{s3}'.isalpha(): {s3.isalpha()}")         # Contains space, so False
print(f"'{s2}'.isalnum(): {s2.isalnum()}")         # All characters are alphanumeric
print(f"'{s3}'.isalnum(): {s3.isalnum()}")         # Contains space, so False
print(f"'{s4}'.isspace(): {s4.isspace()}")         # Contains only whitespace characters

# startswith() and endswith()
filename = "document.pdf"
print(f"'{filename}'.startswith('doc'): {filename.startswith('doc')}")
print(f"'{filename}'.endswith('.pdf'): {filename.endswith('.pdf')}")
```

### 5. Formatted String Literals (f-strings)

Introduced in Python 3.6, f-strings provide a concise and readable way to embed expressions inside string literals. They are prefixed with `f` or `F`.

```python
name = "Alice"
age = 30
profession = "Engineer"

# Traditional string formatting
print("Name: {}, Age: {}, Profession: {}".format(name, age, profession))

# f-string formatting
f_string_example = f"Name: {name}, Age: {age}, Profession: {profession}."
print(f_string_example)

# You can also include expressions directly
item = "Laptop"
price = 1200
tax_rate = 0.08

total_cost = f"The {item} costs ${price:.2f} before tax, and ${price * (1 + tax_rate):.2f} after tax."
print(total_cost)
```
