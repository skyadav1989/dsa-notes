## 🟢 Beginner Level (Q1–35)

### 1. What is SQL?

**Answer:** SQL (Structured Query Language) is used to interact with relational databases. It can create, read, update, and delete data and manage database structures.


````python
SELECT name, salary
FROM employees;
````

### 2. What is a database?

**Answer:** A database is an organized collection of data that can be efficiently stored, accessed, and managed. Examples include MySQL, PostgreSQL, Oracle, and SQL Server.


### 3. What is a table?

**Answer:** A table stores data in rows and columns.


````python
CREATE TABLE employees (
    id INT,
    name VARCHAR(100),
    salary DECIMAL(10,2)
);
````

### 4. What is a row?

**Answer:** A row represents one record in a table.


````python
1 | John | 50000
````

### 5. What is a column?

**Answer:** A column represents a particular attribute of data, such as id, name, or salary.

### 6. What is a primary key?

**Answer:** A primary key uniquely identifies each row. It must be unique and cannot be NULL. A table has one primary-key constraint, which may consist of multiple columns.


````python
CREATE TABLE users (
    id INT PRIMARY KEY,
    name VARCHAR(100)
);
````

### 7. What is a foreign key?

**Answer:** A foreign key creates a relationship between two tables and helps enforce referential integrity.


````python
CREATE TABLE orders (
    id INT PRIMARY KEY,
    user_id INT,
    FOREIGN KEY (user_id) REFERENCES users(id)
);
````

### 8. What is a UNIQUE constraint?

**Answer:** It ensures that values in a column or column set are unique. NULL behavior can vary by database.


````python
email VARCHAR(255) UNIQUE
````

### 9. What is NOT NULL?

**Answer:** It prevents a column from containing NULL values.


````python
name VARCHAR(100) NOT NULL
````

### 10. What is DEFAULT?

**Answer:** It provides a default value when no value is supplied.


````python
status VARCHAR(20) DEFAULT 'active'
````

### 11. Difference between DELETE, DROP, and TRUNCATE?

**Answer:** DELETE removes selected rows and can usually use WHERE. TRUNCATE removes all rows efficiently. DROP removes the table itself. Transaction and identity behavior varies by DBMS.


````python
DELETE FROM users WHERE id = 10;
TRUNCATE TABLE users;
DROP TABLE users;
````

### 12. What is SELECT?

**Answer:** SELECT retrieves data from one or more tables.


````python
SELECT name, salary
FROM employees;
````

### 13. How do you select all columns?

**Answer:** Use SELECT *.


````python
SELECT *
FROM employees;
````

### 14. What is WHERE?

**Answer:** WHERE filters individual rows before grouping or aggregation.


````python
SELECT *
FROM employees
WHERE salary > 50000;
````

### 15. What is ORDER BY?

**Answer:** ORDER BY sorts query results in ascending (ASC) or descending (DESC) order.


````python
SELECT *
FROM employees
ORDER BY salary DESC;
````

### 16. What is GROUP BY?

**Answer:** GROUP BY groups rows with the same grouping values, usually for aggregate calculations.


````python
SELECT department_id, COUNT(*)
FROM employees
GROUP BY department_id;
````

### 17. What is HAVING?

**Answer:** HAVING filters groups after aggregation.


````python
SELECT department_id, COUNT(*)
FROM employees
GROUP BY department_id
HAVING COUNT(*) > 5;
````

### 18. WHERE vs HAVING?

**Answer:** WHERE filters rows before grouping. HAVING filters groups after GROUP BY and aggregation.


````python
SELECT department_id, AVG(salary)
FROM employees
WHERE salary > 30000
GROUP BY department_id
HAVING AVG(salary) > 50000;
````

### 19. What is DISTINCT?

**Answer:** DISTINCT removes duplicate result rows for the selected expression(s).


````python
SELECT DISTINCT department_id
FROM employees;
````

### 20. What is an alias?

**Answer:** An alias gives a temporary name to a column or table, improving readability.


````python
SELECT salary AS employee_salary
FROM employees;

SELECT e.name
FROM employees e;
````

### 21. What is NULL?

**Answer:** NULL represents a missing, unknown, or unavailable value. It is not the same as 0, an empty string, or FALSE.


### 22. How do you check for NULL?

**Answer:** Use IS NULL or IS NOT NULL, not = NULL.


````python
SELECT *
FROM employees
WHERE manager_id IS NULL;
````

### 23. What is COUNT()?

**Answer:** COUNT(*) counts rows. COUNT(column) counts non-NULL values in that column.


````python
SELECT COUNT(*)
FROM employees;

SELECT COUNT(manager_id)
FROM employees;
````

### 24. What are aggregate functions?

**Answer:** Common aggregate functions are COUNT(), SUM(), AVG(), MIN(), and MAX(). They summarize multiple rows.


````python
SELECT AVG(salary)
FROM employees;
````

### 25. What is LIKE?

**Answer:** LIKE performs pattern matching. % means zero or more characters; _ means exactly one character.


````python
SELECT *
FROM employees
WHERE name LIKE 'A%';
````

### 26. What is IN?

**Answer:** IN checks whether a value matches one of several specified values.


````python
SELECT *
FROM employees
WHERE department_id IN (1, 2, 3);
````

### 27. What is BETWEEN?

**Answer:** BETWEEN tests whether a value is within an inclusive range for standard SQL comparisons.


````python
SELECT *
FROM employees
WHERE salary BETWEEN 40000 AND 80000;
````

### 28. UNION vs UNION ALL?

**Answer:** UNION combines compatible result sets and removes duplicates. UNION ALL combines them while retaining duplicates and is often faster.


````python
SELECT city FROM customers
UNION
SELECT city FROM suppliers;

SELECT city FROM customers
UNION ALL
SELECT city FROM suppliers;
````

### 29. What is a JOIN?

**Answer:** A JOIN combines rows from multiple tables based on a related condition.


````python
SELECT e.name, d.name
FROM employees e
JOIN departments d
  ON e.department_id = d.id;
````

### 30. What is INNER JOIN?

**Answer:** INNER JOIN returns only rows that have a matching row on both sides.


````python
SELECT *
FROM employees e
INNER JOIN departments d
  ON e.department_id = d.id;
````

### 31. What is LEFT JOIN?

**Answer:** LEFT JOIN returns all rows from the left table and matching rows from the right table. Unmatched right-side columns are NULL.


````python
SELECT *
FROM employees e
LEFT JOIN departments d
  ON e.department_id = d.id;
````

### 32. What is RIGHT JOIN?

**Answer:** RIGHT JOIN returns all rows from the right table and matching rows from the left table. Unmatched left-side columns are NULL.


````python
SELECT *
FROM employees e
RIGHT JOIN departments d
  ON e.department_id = d.id;
````

### 33. What is a self join?

**Answer:** A self join joins a table to itself, commonly for employee-manager relationships.


````python
SELECT
    e.name AS employee,
    m.name AS manager
FROM employees e
LEFT JOIN employees m
  ON e.manager_id = m.id;
````

### 34. What is a subquery?

**Answer:** A subquery is a query nested inside another SQL statement.


````python
SELECT *
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
````

### 35. What is a view?

**Answer:** A view is a virtual table defined by a query. Its exact materialization and update behavior depends on the DBMS.


````python
CREATE VIEW employee_view AS
SELECT id, name, salary
FROM employees;
````

## 🟡 Advanced Level (Q36–70)

### 36. What is an index?

**Answer:** An index is a data structure that can speed up data retrieval, usually at the cost of additional storage and slower writes.


````python
CREATE INDEX idx_employee_email
ON employees(email);
````

### 37. What is a composite index?

**Answer:** A composite index contains multiple columns. Column order matters for how a B-tree index can be used.


````python
CREATE INDEX idx_employee_dept_salary
ON employees(department_id, salary);
````

### 38. What is a clustered index?

**Answer:** A clustered index determines how table rows are physically/logically organized in systems that support this concept. For example, SQL Server supports clustered indexes; InnoDB uses the primary key as the clustered index.


### 39. What is a non-clustered index?

**Answer:** A non-clustered index is a separate index structure containing indexed values and references to the corresponding table rows.


### 40. What is normalization?

**Answer:** Normalization organizes relational data to reduce unnecessary duplication and improve data integrity. Common forms include 1NF, 2NF, 3NF, BCNF, 4NF, and 5NF.


### 41. What is 1NF?

**Answer:** First Normal Form requires atomic column values and no repeating groups.


````python
-- Bad design:
id | phones
1  | 9999,8888

-- Better:
id | phone
1  | 9999
1  | 8888
````

### 42. What is 2NF?

**Answer:** A table is in 2NF when it is in 1NF and every non-key attribute depends on the entire candidate key. It mainly matters when a key is composite.


### 43. What is 3NF?

**Answer:** A table is in 3NF when it is in 2NF and non-key attributes do not depend transitively on another non-key attribute.


### 44. What is denormalization?

**Answer:** Denormalization intentionally introduces controlled redundancy to improve read performance or simplify reporting queries.


### 45. What is a transaction?

**Answer:** A transaction is a logical unit of database work whose operations are committed or rolled back according to transaction semantics.


````python
BEGIN;

UPDATE accounts
SET balance = balance - 100
WHERE id = 1;

UPDATE accounts
SET balance = balance + 100
WHERE id = 2;

COMMIT;
````

### 46. What is ACID?

**Answer:** Atomicity: all-or-nothing. Consistency: constraints/rules remain valid. Isolation: concurrent transactions are isolated according to the isolation level. Durability: committed changes survive failures.


### 47. What is COMMIT?

**Answer:** COMMIT makes the current transaction's changes permanent, subject to the DBMS transaction model.


````python
COMMIT;
````

### 48. What is ROLLBACK?

**Answer:** ROLLBACK undoes uncommitted transactional changes.


````python
ROLLBACK;
````

### 49. What is a SAVEPOINT?

**Answer:** A SAVEPOINT marks a point inside a transaction so you can roll back part of the transaction without necessarily rolling back everything.


````python
SAVEPOINT sp1;
ROLLBACK TO SAVEPOINT sp1;
````

### 50. What is a stored procedure?

**Answer:** A stored procedure is a database-side program containing SQL statements and procedural logic. Syntax and capabilities vary by DBMS.


### 51. What is a function?

**Answer:** A database function accepts inputs and returns a value or table/result depending on the DBMS. Functions can be built-in or user-defined.


````python
SELECT UPPER(name)
FROM employees;
````

### 52. Stored procedure vs function?

**Answer:** Procedures are commonly used for operations/workflows and may return result sets. Functions generally return a value/table and can be used in expressions depending on the DBMS. Exact rules vary by database.


### 53. What is a CTE?

**Answer:** CTE (Common Table Expression) defines a named temporary result set for one statement, improving readability and enabling recursive queries.


````python
WITH high_salary AS (
    SELECT *
    FROM employees
    WHERE salary > 100000
)
SELECT *
FROM high_salary;
````

### 54. What is a recursive CTE?

**Answer:** A recursive CTE references itself and is useful for hierarchical data such as organizational charts or category trees.


````python
WITH RECURSIVE employee_tree AS (
    SELECT id, name, manager_id
    FROM employees
    WHERE manager_id IS NULL

    UNION ALL

    SELECT e.id, e.name, e.manager_id
    FROM employees e
    JOIN employee_tree t
      ON e.manager_id = t.id
)
SELECT *
FROM employee_tree;
````

### 55. What is a window function?

**Answer:** A window function calculates across related rows without collapsing them into one row per group.


````python
SELECT
    name,
    salary,
    AVG(salary) OVER (
        PARTITION BY department_id
    ) AS dept_avg
FROM employees;
````

### 56. RANK vs DENSE_RANK vs ROW_NUMBER?

**Answer:** ROW_NUMBER gives every row a unique sequence. RANK gives tied rows the same rank and leaves gaps. DENSE_RANK gives tied rows the same rank without gaps.


````python
-- Salaries: 100, 100, 90, 80
-- ROW_NUMBER: 1, 2, 3, 4
-- RANK:       1, 1, 3, 4
-- DENSE_RANK: 1, 1, 2, 3
````

### 57. What is PARTITION BY?

**Answer:** PARTITION BY divides rows into independent groups for a window function.


````python
AVG(salary) OVER (
    PARTITION BY department_id
)
````

### 58. What is an execution plan?

**Answer:** An execution plan shows how the database optimizer intends to execute a query, including scans/seeks, joins, sorts, estimates, and costs. EXPLAIN is commonly used, with syntax varying by DBMS.


````python
EXPLAIN
SELECT *
FROM employees
WHERE email = 'a@example.com';
````

### 59. How do you optimize a slow query?

**Answer:** Measure first. Inspect the execution plan, indexes, predicates, joins, statistics, cardinality estimates, sorting, locking, and I/O. Avoid unnecessary columns and rows, then benchmark changes.


### 60. Why can SELECT * be bad?

**Answer:** It may read unnecessary columns, increase network traffic, reduce opportunities for covering indexes, and make applications more fragile when schemas change.


````python
SELECT id, name, salary
FROM employees;
````

### 61. What is a covering index?

**Answer:** A covering index contains all columns needed for a query, allowing the database to satisfy the query from the index without fetching the base table row in many cases.


````python
CREATE INDEX idx_emp_dept
ON employees(department_id, salary, name);
````

### 62. What is cardinality?

**Answer:** Cardinality can mean the number of distinct values in a column or the number of rows produced by an operation, depending on context. Index selectivity often benefits from higher distinctness.


### 63. What is selectivity?

**Answer:** Selectivity describes how strongly a predicate narrows the candidate rows. A predicate matching very few rows is highly selective.


````python
WHERE employee_id = 123
````

### 64. What is a deadlock?

**Answer:** A deadlock occurs when concurrent transactions wait on resources held by each other, so neither can proceed. Databases typically detect and abort one transaction.


````python
-- Concept:
-- Tx A locks Row 1, waits for Row 2
-- Tx B locks Row 2, waits for Row 1
````

### 65. How can you reduce deadlocks?

**Answer:** Keep transactions short, access resources in a consistent order, use appropriate indexes, avoid unnecessary locks, avoid user interaction inside transactions, and retry transactions aborted due to deadlocks where appropriate.


### 66. What is a database constraint?

## 🔴 Expert Level (Q71–100)

### 71. What is transaction isolation?

**Answer:** Transaction isolation controls how concurrent transactions can observe each other's changes. Common levels include READ UNCOMMITTED, READ COMMITTED, REPEATABLE READ, and SERIALIZABLE. Implementations vary by DBMS.


### 72. What is a dirty read?

**Answer:** A dirty read occurs when a transaction reads data written by another transaction before that transaction commits, and the writer may later roll back.


### 73. What is a non-repeatable read?

**Answer:** A transaction reads the same row twice and gets different committed values because another transaction updated the row between reads.


### 74. What is a phantom read?

**Answer:** A transaction repeats a range query and sees additional or missing matching rows because another transaction inserted or deleted rows in that range.


### 75. What is MVCC?

**Answer:** MVCC (Multi-Version Concurrency Control) maintains multiple row versions so readers can often see a consistent snapshot while writers continue. Implementations differ by DBMS.


### 76. What is optimistic locking?

**Answer:** Optimistic locking assumes conflicts are uncommon. A version column or timestamp is checked during UPDATE; if no row is updated, another transaction likely changed the record.


````python
UPDATE accounts
SET balance = 900,
    version = 6
WHERE id = 1
  AND version = 5;
````

### 77. What is pessimistic locking?

**Answer:** Pessimistic locking acquires a lock before modification so conflicting transactions may have to wait, depending on the DBMS and lock mode.


````python
SELECT *
FROM accounts
WHERE id = 1
FOR UPDATE;
````

### 78. What is a materialized view?

**Answer:** A materialized view stores the result of a query physically. It can accelerate expensive reads but requires refresh/maintenance and can become stale.


### 79. View vs materialized view?

**Answer:** A normal view usually stores a query definition and computes results when queried. A materialized view stores query results and may need refreshing. Exact behavior varies by DBMS.


### 80. What is query pagination?

**Answer:** Pagination limits the amount of data returned per page. OFFSET/LIMIT is simple but large offsets can become expensive.


````python
SELECT *
FROM products
ORDER BY id
LIMIT 20 OFFSET 100;
````

### 81. OFFSET pagination vs keyset pagination?

**Answer:** OFFSET pagination skips rows and can degrade for large offsets. Keyset pagination uses a stable ordering key and a continuation predicate, often scaling better for large datasets.


````python
SELECT *
FROM products
WHERE id > 100
ORDER BY id
LIMIT 20;
````

### 82. How do you find the second-highest salary?

**Answer:** For the second distinct-highest salary, find the maximum salary below the overall maximum.


````python
SELECT MAX(salary)
FROM employees
WHERE salary < (
    SELECT MAX(salary)
    FROM employees
);
````

### 83. Find the Nth highest salary.

**Answer:** Use DENSE_RANK when you want the Nth distinct salary. Use ROW_NUMBER if you want the Nth physical row after ordering.


````python
SELECT salary
FROM (
    SELECT
        salary,
        DENSE_RANK() OVER (
            ORDER BY salary DESC
        ) AS rnk
    FROM employees
) x
WHERE rnk = 5;
````

### 84. Find duplicate emails.

**Answer:** Group by email and keep groups with more than one row.


````python
SELECT email, COUNT(*) AS cnt
FROM users
GROUP BY email
HAVING COUNT(*) > 1;
````

### 85. Find employees earning more than their manager.

**Answer:** Self-join employees to managers and compare salaries.


````python
SELECT e.name
FROM employees e
JOIN employees m
  ON e.manager_id = m.id
WHERE e.salary > m.salary;
````

### 86. Find the highest-paid employee in each department.

**Answer:** Use ROW_NUMBER for exactly one employee per department; use RANK if all tied highest earners should be returned.


````python
SELECT *
FROM (
    SELECT
        e.*,
        ROW_NUMBER() OVER (
            PARTITION BY department_id
            ORDER BY salary DESC
        ) AS rn
    FROM employees e
) x
WHERE rn = 1;
````

### 87. Find the top 3 employees per department.

**Answer:** Use a ranking window function partitioned by department. DENSE_RANK returns the top three distinct salary ranks and can return ties.


````python
SELECT *
FROM (
    SELECT
        e.*,
        DENSE_RANK() OVER (
            PARTITION BY department_id
            ORDER BY salary DESC
        ) AS rnk
    FROM employees e
) x
WHERE rnk <= 3;
````

### 88. How do you find customers who never placed an order?

**Answer:** NOT EXISTS is a clear and NULL-safe way to express this anti-join.


````python
SELECT c.*
FROM customers c
WHERE NOT EXISTS (
    SELECT 1
    FROM orders o
    WHERE o.customer_id = c.id
);
````

### 89. NOT IN vs NOT EXISTS?

**Answer:** NOT IN can produce surprising results if the subquery contains NULL because of SQL's three-valued logic. NOT EXISTS is often safer for anti-join logic.


````python
SELECT c.*
FROM customers c
WHERE NOT EXISTS (
    SELECT 1
    FROM orders o
    WHERE o.customer_id = c.id
);
````

### 90. What is a correlated subquery?

**Answer:** A correlated subquery references a value from the outer query, so its logic depends on the current outer row.


````python
SELECT e.*
FROM employees e
WHERE salary > (
    SELECT AVG(e2.salary)
    FROM employees e2
    WHERE e2.department_id = e.department_id
);
````

### 91. What is a SARGable query?

**Answer:** A SARGable predicate is written in a form that allows an index to be used efficiently where appropriate. Wrapping an indexed column in a function can prevent efficient index access in many cases.


````python
-- Often more index-friendly:
WHERE created_at >= '2026-01-01'

-- May be less index-friendly:
WHERE YEAR(created_at) = 2026
````

### 92. Why can functions on indexed columns hurt performance?

**Answer:** Applying a function to an indexed column can make a normal index difficult to use efficiently because the database must evaluate the function for many rows. Functional indexes or generated columns may help, depending on the DBMS.


````python
WHERE LOWER(email) = 'john@example.com'
````

### 93. What is a composite index and why does column order matter?

**Answer:** Composite indexes contain multiple columns. For B-tree indexes, the leading column(s) are especially important for efficient matching; the optimizer's exact choices depend on the query and DBMS.


````python
CREATE INDEX idx_orders
ON orders(customer_id, status, created_at);
````

### 94. Why might the database ignore an index?

**Answer:** Possible reasons include low selectivity, too many matching rows, stale statistics, inaccurate cardinality estimates, functions/casts on columns, leading wildcards, or a table scan simply being cheaper.


````python
-- Example:
WHERE name LIKE '%abc'
````

### 95. What is an index scan vs index seek?

**Answer:** An index seek navigates to a relevant part of an index. An index scan reads a larger portion or all of an index. A scan is not automatically bad if a large percentage of rows is needed.


### 96. What is a hash join?

**Answer:** A hash join builds a hash structure from one input and probes it with the other, commonly for equality joins. The optimizer chooses it when its cost model favors it.


````python
SELECT *
FROM orders o
JOIN customers c
  ON o.customer_id = c.id;
````

### 97. What is a nested-loop join?

**Answer:** A nested-loop join processes rows from one input and looks for matching rows in the other. It can be efficient when the outer input is small and the inner side has a useful index.


### 98. What is a merge join?

**Answer:** A merge join combines two inputs that are ordered on the join keys by advancing through them in order. It can be efficient when suitable sorted inputs are available.


### 99. What is partitioning?

**Answer:** Partitioning divides a large table or index into smaller partitions while exposing it as one logical table. Common strategies include RANGE, LIST, and HASH. Partition pruning can reduce work for queries that filter on the partition key.


### 100. How would you optimize a query running on a 500-million-row table?

**Answer:** Measure before changing anything. Inspect the actual execution plan, row estimates, I/O, joins, sorting, predicates, indexes, statistics, data distribution, locking, and partition pruning. Make predicates SARGable, return only needed data, consider keyset pagination for large result sets, and evaluate pre-aggregation/materialized views for analytical workloads. Benchmark changes against production-like data and verify their effect on writes and concurrency.
