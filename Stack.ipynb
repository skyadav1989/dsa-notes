# Stack Data Structure

A **stack** is a linear data structure that follows a particular order in which operations are performed. The order is **LIFO (Last In, First Out)**. Think of a stack of plates: you can only add a new plate to the top, and you can only remove the topmost plate.

### Key Operations:
*   **Push**: Adds an item to the top of the stack.
*   **Pop**: Removes the item from the top of the stack.
*   **Peek (or Top)**: Returns the top item without removing it.
*   **Is Empty**: Checks if the stack is empty.
*   **Size**: Returns the number of items in the stack.

## Python Implementation of a Stack

We can implement a stack in Python using a simple list. Lists provide efficient `append()` for pushing and `pop()` for popping from the end, which mimics the LIFO behavior.

```python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            return "Stack is empty"
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return "Stack is empty"
        return self.items[-1]

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)
```

## Demonstrating Stack Operations

Let's create an instance of our `Stack` and perform some operations.

```python
# Create a new stack
my_stack = Stack()
print(f"Initial stack: {my_stack}")
```

### 1. Push Operation

`push(item)` adds an element to the top of the stack.

```python
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
print(f"Stack after pushing 10, 20, 30: {my_stack}")
```

### 2. Pop Operation

`pop()` removes and returns the topmost element from the stack. Because it's LIFO, the last element pushed (30) will be the first one popped.

```python
popped_item = my_stack.pop()
print(f"Popped item: {popped_item}")
print(f"Stack after one pop: {my_stack}")

popped_item = my_stack.pop()
print(f"Popped item: {popped_item}")
print(f"Stack after second pop: {my_stack}")
```

### 3. Peek Operation

`peek()` (or `top()`) returns the topmost element without removing it from the stack.

```python
#my_stack.push(40)
top_item = my_stack.peek()
print(f"Top item (peek): {top_item}")
print(f"Stack after peek: {my_stack}")
```

### 4. Is Empty Operation

`is_empty()` checks if the stack contains any elements.

```python
print(f"Is stack empty? {my_stack.is_empty()}")

# Pop remaining item to make it empty
my_stack.pop()
print(f"Stack after popping remaining item: {my_stack}")
print(f"Is stack empty? {my_stack.is_empty()}")
```

### 5. Size Operation

`size()` returns the number of elements currently in the stack.

```python
my_stack.push('A')
my_stack.push('B')
my_stack.push('C')
print(f"Stack: {my_stack}")
print(f"Current size of the stack: {my_stack.size()}")

my_stack.pop()
print(f"Stack after one pop: {my_stack}")
print(f"Current size of the stack: {my_stack.size()}")
```

### Handling Empty Stack Edge Cases

Attempting to `pop()` or `peek()` from an empty stack should be handled gracefully.

```python
empty_stack = Stack()
print(f"Attempting to pop from an empty stack: {empty_stack.pop()}")
print(f"Attempting to peek from an empty stack: {empty_stack.peek()}")
```
