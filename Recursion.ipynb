## Recursion in Data Structures and Algorithms (DSA)

Recursion is a programming technique where a function calls itself to solve a problem. It's a powerful tool for solving problems that can be broken down into smaller, self-similar subproblems.

### 1. Basics of Recursion

Every recursive function needs two main components:

1.  **Base Case**: This is the condition that stops the recursion. Without a base case, the function would call itself indefinitely, leading to a stack overflow.
2.  **Recursive Step**: This is where the function calls itself with a modified (usually smaller) input, moving closer to the base case.

### 2. Examples of Recursion

#### Factorial

The factorial of a non-negative integer `n` is the product of all positive integers less than or equal to `n`. The factorial of 0 is 1. We can define it recursively as:
*   `n! = n * (n-1)!` for `n > 0`
*   `0! = 1` (Base Case)

```python
def factorial(n):
    if n == 0:  # Base Case
        return 1
    else:  # Recursive Step
        return n * factorial(n - 1)

print(f"Factorial of 5: {factorial(5)}")
print(f"Factorial of 0: {factorial(0)}")
```

#### Fibonacci Sequence

The Fibonacci sequence is a series of numbers where each number is the sum of the two preceding ones, usually starting with 0 and 1. We can define it recursively as:
*   `F(n) = F(n-1) + F(n-2)` for `n > 1`
*   `F(0) = 0` (Base Case)
*   `F(1) = 1` (Base Case)

```python
def fibonacci(n):
    if n <= 1:  # Base Cases
        return n
    else:  # Recursive Step
        return fibonacci(n - 1) + fibonacci(n - 2)

# Print the first 10 Fibonacci numbers
print("Fibonacci sequence (first 10 numbers):")
for i in range(10):
    print(fibonacci(i), end=" ")
```

### 3. Tail vs Non-Tail Recursion

This distinction is important in some programming languages because tail-recursive calls can be optimized by compilers/interpreters to avoid growing the call stack, effectively turning recursion into iteration.

*   **Tail Recursion**: A function is tail-recursive if the recursive call is the *last operation* performed in the function. Nothing is done with the result of the recursive call after it returns.
*   **Non-Tail Recursion**: If there are operations performed after the recursive call returns (e.g., multiplication, addition), then it is non-tail recursive. The `factorial` and `fibonacci` examples above are non-tail recursive because they perform an operation (multiplication or addition) after the recursive call.

Python does *not* perform Tail Call Optimization (TCO), so the performance benefits of tail recursion are not realized in Python.

#### Tail-Recursive Factorial (Conceptual Example for Python)

While Python doesn't optimize tail calls, we can still write a function in a tail-recursive style by passing an accumulator argument.

```python
def factorial_tail_recursive(n, accumulator=1):
    if n == 0:  # Base Case
        return accumulator
    else:  # Recursive Step (last operation is the recursive call)
        return factorial_tail_recursive(n - 1, accumulator * n)

print(f"Factorial of 5 (tail-recursive): {factorial_tail_recursive(5)}")
print(f"Factorial of 0 (tail-recursive): {factorial_tail_recursive(0)}")
```

### 4. Tower of Hanoi

The Tower of Hanoi is a mathematical puzzle. It consists of three rods and a number of disks of different sizes, which can slide onto any rod. The puzzle starts with the disks in a neat stack in ascending order of size on one rod, the smallest at the top, thus making a conical shape.

The objective of the puzzle is to move the entire stack from the source rod to the destination rod, obeying the following simple rules:
1.  Only one disk can be moved at a time.
2.  Each move consists of taking the upper disk from one of the stacks and placing it on top of another stack or on an empty rod.
3.  No disk may be placed on top of a smaller disk.

This problem is famously solved using recursion.

```python
def tower_of_hanoi(n, source, auxiliary, destination):
    if n == 1:
        print(f"Move disk 1 from {source} to {destination}")
        return
    # Move n-1 disks from source to auxiliary, using destination as temporary
    tower_of_hanoi(n - 1, source, destination, auxiliary)
    # Move the nth disk from source to destination
    print(f"Move disk {n} from {source} to {destination}")
    # Move n-1 disks from auxiliary to destination, using source as temporary
    tower_of_hanoi(n - 1, auxiliary, source, destination)

print("\nTower of Hanoi for 3 disks:")
tower_of_hanoi(3, 'A', 'B', 'C')
```
