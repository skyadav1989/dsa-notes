# Queue Data Structure

A **Queue** is a linear data structure that follows a particular order in which the operations are performed. The order is **First In, First Out (FIFO)**. This means the element that is inserted first is the first one to be removed.

Imagine a line at a ticket counter: the first person to join the line is the first person to get a ticket and leave the line.

### Core Operations of a Queue:

*   **Enqueue (or Push)**: Adds an element to the rear (end) of the queue.
*   **Dequeue (or Pop)**: Removes an element from the front (beginning) of the queue.
*   **Peek (or Front)**: Returns the element at the front of the queue without removing it.
*   **isEmpty**: Checks if the queue is empty.
*   **Size**: Returns the number of elements in the queue.

## 1. Basic Queue Implementation using a Python List

We can implement a simple queue using a Python list. While lists are flexible, using `append()` for enqueue and `pop(0)` for dequeue can be inefficient for very large queues because `pop(0)` requires shifting all subsequent elements.

### Initialization

```python
# Initialize an empty queue using a list
queue = []
print(f"Initialized queue: {queue}")
```

### Enqueue (Adding elements)

Elements are added to the `rear` of the queue. In a list-based implementation, we typically use `append()` to add elements to the end of the list.

```python
queue.append('Alice') # Enqueue 'Alice'
queue.append('Bob')   # Enqueue 'Bob'
queue.append('Charlie') # Enqueue 'Charlie'
print(f"Queue after enqueue operations: {queue}")
```

### Dequeue (Removing elements)

Elements are removed from the `front` of the queue. According to the FIFO principle, the first element added will be the first one removed. In a list, this corresponds to `pop(0)`.

```python
if queue:
    dequeued_item = queue.pop(0) # Dequeue the front element
    print(f"Dequeued: {dequeued_item}")
    print(f"Queue after first dequeue: {queue}")
else:
    print("Queue is empty, cannot dequeue.")

if queue:
    dequeued_item = queue.pop(0)
    print(f"Dequeued: {dequeued_item}")
    print(f"Queue after second dequeue: {queue}")
else:
    print("Queue is empty, cannot dequeue.")
```

### Peek (Viewing the front element)

To look at the element at the front of the queue without removing it, we can access the first element of the list (`queue[0]`).

```python
if queue:
    front_item = queue[0] # Peek at the front element
    print(f"Front item (peek): {front_item}")
    print(f"Queue after peek: {queue}") # Queue remains unchanged
else:
    print("Queue is empty, no front item to peek.")
```

### isEmpty (Checking if the queue is empty)

We can check if the queue is empty by simply checking the length of the list (`len(queue) == 0`). A boolean conversion of the list also works (`not queue`).

```python
print(f"Is the queue empty? {not queue}") # Check if empty

# Let's empty the queue
while queue:
    queue.pop(0)
print(f"Queue after emptying: {queue}")
print(f"Is the queue empty now? {not queue}")
```

### Size (Getting the number of elements)

The number of elements in the queue can be found using the `len()` function on the list.

```python
queue.append('First')
queue.append('Second')
print(f"Current queue: {queue}")
print(f"Size of the queue: {len(queue)}")
```

## 2. Queue Implementation using `collections.deque`

For a more efficient queue implementation in Python, especially for large datasets, the `collections.deque` (double-ended queue) class is recommended. It provides O(1) (constant time) complexity for adding and removing elements from both ends, unlike lists where `pop(0)` is O(n).

```python
from collections import deque

# Initialize a deque as a queue
queue_deque = deque()
print(f"Initialized deque queue: {queue_deque}")

# Enqueue: use append()
queue_deque.append('Apple')
queue_deque.append('Banana')
queue_deque.append('Cherry')
print(f"Deque queue after enqueue: {queue_deque}")

# Dequeue: use popleft()
if queue_deque:
    dequeued_item_deque = queue_deque.popleft()
    print(f"Dequeued from deque: {dequeued_item_deque}")
    print(f"Deque queue after dequeue: {queue_deque}")
else:
    print("Deque queue is empty, cannot dequeue.")

# Peek (front element): access with [0]
if queue_deque:
    front_item_deque = queue_deque[0]
    print(f"Front item (peek) from deque: {front_item_deque}")
    print(f"Deque queue after peek: {queue_deque}") # Unchanged
else:
    print("Deque queue is empty, no front item to peek.")

# isEmpty: check length or boolean conversion
print(f"Is the deque queue empty? {not queue_deque}")

# Size: use len()
print(f"Size of the deque queue: {len(queue_deque)}")

# Empty the deque
queue_deque.clear()
print(f"Deque queue after clear: {queue_deque}")
print(f"Is the deque queue empty now? {not queue_deque}")
```
